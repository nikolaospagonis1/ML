## 1.1 Why Python for Machine Learning?

Python is a **high-level, interpreted, general-purpose** programming language.

| Property | What it means | Why it matters for ML |
|---|---|---|
| High-level | You describe *what* to do, not how memory is managed | You focus on the model, not on the machine |
| Interpreted | Code runs line by line, no compile step | Fast experimentation in notebooks |
| Dynamically typed | A variable's type is decided at run time | Less boilerplate code |
| Huge ecosystem | NumPy, pandas, scikit-learn, PyTorch, TensorFlow… | Almost every ML tool has a Python interface |

A key idea: **Python itself is slow, but its ML libraries are fast** because they are written in C/C++/CUDA underneath. Python is the "glue" that lets you drive that fast code easily. You will see this clearly in Module 7 (NumPy).

## 1.3 Variables and assignment

A **variable** is a *name* that refers to a *value* stored in memory. We create one with the assignment operator `=`.

> Read `x = 5` as "**x refers to 5**", not "x equals 5". The `=` sign does *assignment*; equality testing uses `==`.

````python
learning_rate = 0.01
epochs = 100
model_name = "logistic_regression "

print(learning_rate, epochs, model_name)
````

Variables can be **re-assigned**, even to a value of a different type (dynamic typing):

````python
x = 10
print(x)
x = "now I am a string"
print(x)
````

**Naming rules and conventions**

* Names may contain letters, digits and `_`, but **cannot start with a digit**.
* They are **case-sensitive**: `Accuracy` and `accuracy` are different variables.
* Reserved keywords (`if`, `for`, `class`, `lambda`, …) cannot be used.
* Convention (PEP 8): use `snake_case` for variables and functions, `CamelCase` for classes, `UPPER_CASE` for constants.

````python
# Multiple assignment and swapping
a, b = 1, 2
a, b = b, a          # swap without a temporary variable
print(a, b)

# Constants are a convention only - Python does not enforce them
RANDOM_SEED = 42
````

## 1.4 Core data types

Every value has a **type**. Use `type()` to inspect it.

| Type | Example | Description |
|---|---|---|
| `int` | `42`, `-7` | Whole numbers, arbitrary size |
| `float` | `3.14`, `1e-3` | Real numbers (64-bit floating point) |

````python
print(type(42))
print(type(3.14))
print(type(True))
print(type("text"))
print(type(None))
````

### Integers have unlimited precision

````python
2 ** 200
````

### Floats are approximations

Floats are stored in binary, so many decimal numbers cannot be represented exactly. This is **not** a Python bug — it happens in every language and matters in ML when comparing numbers.

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)

import math
print(math.isclose(0.1 + 0.2, 0.3))   # the correct way to compare floats
````

Scientific notation is common in ML (learning rates, tolerances):

````python
tol = 1e-6       # 0.000001
big = 3.2e5      # 320000.0
print(tol, big)
````

## 1.5 Operators

### Arithmetic

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` `*` | add, subtract, multiply | `7 * 2` | `14` |
| `/` | true division (always float) | `7 / 2` | `3.5` |
| `//` | floor division | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | power | `2 ** 3` | `8` |

````python
print(7 / 2, 7 // 2, 7 % 2, 2 ** 3)
print(-7 // 2)   # floor division rounds DOWN (towards -infinity), giving -4
````

Precedence follows mathematics: `**` first, then `* / // %`, then `+ -`. **Use parentheses when in doubt.**

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

Augmented assignment is a shorthand you will use constantly in training loops:

````python
total_loss = 0.0
total_loss += 0.5    # same as total_loss = total_loss + 0.5
total_loss += 0.25
total_loss *= 2
total_loss
````

### Comparison operators

They return a `bool`: `==`, `!=`, `<`, `<=`, `>`, `>=`. Python allows **chained comparisons**:

````python
accuracy = 0.87
print(accuracy > 0.8)
print(0.8 < accuracy <= 0.9)   # same as (0.8 < accuracy) and (accuracy <= 0.9)
````

### Logical operators: `and`, `or`, `not`

````python
is_trained = True
has_data = False
print(is_trained and has_data)
print(is_trained or has_data)
print(not has_data)
````

**Truthiness.** In a boolean context, the values `0`, `0.0`, `""`, `None`, and empty containers (`[]`, `{}`) count as `False`; everything else counts as `True`.

````python
print(bool(0), bool(""), bool(None), bool([]))
print(bool(5), bool("hi"), bool([0]))
````

## 1.6 Type conversion

````python
print(int("42") + 1)        # str -> int
print(float("3.5") * 2)     # str -> float
print(str(100) + " epochs") # int -> str
print(int(3.99))            # truncates towards zero, does not round!
print(round(3.99))          # rounds
print(round(2.5), round(3.5))  # "banker's rounding": ties go to the even number
````

Mixing types: `int` and `float` combine fine (result is `float`), but `str` + `int` is an **error**:

````python
try:
    "Epoch " + 5
except TypeError as e:
    print("TypeError:", e)

print("Epoch " + str(5))   # fix: convert explicitly
````

### f-strings: the best way to format output

Put an `f` before the quotes and write expressions inside `{}`. After a colon you can give a **format specification**.

````python
epoch = 7
loss = 0.123456789
acc = 0.91234

print(f"Epoch {epoch}: loss={loss:.4f}, accuracy={acc:.1%}")
print(f"{'model':<10}|{'score':>8}")     # < left-align, > right-align in a field of width N
print(f"{'SVM':<10}|{0.8731:>8.3f}")
print(f"{1234567.891:,.2f}")             # thousands separator
print(f"{epoch=}, {loss=:.2f}")          # '=' shows name and value: handy for debugging
````

## 1.8 Strings

A string is an **immutable sequence of characters**. Single `'...'` and double `"..."` quotes are equivalent; triple quotes `"""..."""` span multiple lines.

````python
s1 = 'single'
s2 = "double, with an apostrophe: it's fine"
s3 = """multi
line"""
path = r"C:\data\new"   # raw string: backslashes are not escapes
print(s2)
print(s3)
print(path)
print("tab:\there, newline:\nhere")
````

### Indexing and slicing

Positions start at **0**. Negative indices count from the end. A slice `s[start:stop:step]` includes `start` but **excludes `stop`**.

```text
 string:   P   y   t   h   o   n
 index:    0   1   2   3   4   5
 negative:-6  -5  -4  -3  -2  -1
```

````python
s = "Python"
print(s[0], s[-1])     # first and last character
print(s[0:2])          # 'Py'  (positions 0 and 1)
print(s[2:])           # from position 2 to the end
print(s[:3])           # the first 3 characters
print(s[::2])          # every second character
print(s[::-1])         # reversed
print(len(s))
````

Slicing is a pattern you will reuse for lists, NumPy arrays and pandas objects, so make sure you understand it now.

### Strings are immutable

````python
s = "Python"
try:
    s[0] = "J"
except TypeError as e:
    print("TypeError:", e)

s = "J" + s[1:]   # create a NEW string instead
print(s)
````

### Common string methods

Methods are functions attached to an object, called with `object.method()`. String methods **return new strings**; they never change the original.

````python
text = "  Machine Learning is FUN  "
print(text.strip())                 # remove surrounding whitespace
print(text.lower())
print(text.upper())
print(text.strip().title())
print(text.replace("FUN", "powerful"))
print(text.count("n"))
print(text.find("Learning"))        # index of first match, -1 if absent
print("Learning" in text)           # membership test
print(text.strip().startswith("Machine"))
````

`split` and `join` are inverses and are used constantly when processing text data:

````python
line = "sepal_length,sepal_width,petal_length,species"
columns = line.split(",")
print(columns)
print(" | ".join(columns))
print("The quick brown fox".split())   # default: split on any whitespace
````

String checks:

````python
print("123".isdigit(), "abc".isalpha(), "abc123".isalnum(), "   ".isspace())
````
