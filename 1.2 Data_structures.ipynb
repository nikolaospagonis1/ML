## 2.2 Lists

A list is an **ordered, mutable** collection. Items can be of any type.

```python
losses = [0.92, 0.71, 0.55, 0.43]
mixed = [1, "two", 3.0, True, None, [5, 6]]
empty = []

print(losses)
print(len(losses))
print(mixed[5])
```

### Indexing and slicing (same rules as strings)

```python
losses = [0.92, 0.71, 0.55, 0.43, 0.38, 0.35]
print("1st:",losses[-5:-1])
print("2d",losses[0:-1])
print("3d",losses[1:4])
print("4th",losses[::2])  #[0][2][4]
print("5th",losses[-3:])   # last three values - e.g. to check convergence
```

### Lists are mutable

```python
layers = [64, 128, 256]
layers[0] = 32                # replace an element
print(layers)

layers.append(512)            # add one item at the end
layers.extend([1024, 2048])   # add many items
layers.insert(0, 16)          # insert at a position
print(layers)

last = layers.pop()           # remove AND return the last item
print("popped:", last, "->", layers)
layers.remove(256)            # remove the first occurrence of a VALUE
del layers[0]                 # delete by INDEX
print(layers)
```

### Other useful list operations

```python
scores = [0.81, 0.93, 0.77, 0.93, 0.88]
print(len(scores), min(scores), max(scores), sum(scores))
print("mean:", sum(scores) / len(scores))
print(scores.index(0.93))     # position of the first match
print(scores.count(0.93))     # how many times it appears
print(0.77 in scores)         # membership test

print(sorted(scores))                 # returns a NEW sorted list
print(sorted(scores, reverse=True))
scores.sort()                         # sorts IN PLACE and returns None
print(scores)

print([1, 2] + [3, 4])   # concatenation
print([0] * 5)           # repetition: a quick way to initialise
```

> ⚠️ **Common bug:** `scores = scores.sort()` sets `scores` to `None`, because `.sort()` modifies the list in place and returns nothing. Use either `scores.sort()` or `scores = sorted(scores)`.

### Unpacking

```python
shape = [150, 4]
n_rows, n_cols = shape
print(n_rows, n_cols)

first, *middle, last = [1, 2, 3, 4, 5]   # the * collects the rest
print(first, middle, last)
```

## 2.3 Mutability, references and copies — very important!

In Python, a variable holds a **reference** to an object. Assigning one variable to another does **not** copy the object; both names now refer to the *same* object.

```python
a = [1, 2, 3]
b = a            # b refers to the SAME list
b.append(4)
print("a:", a)   # a changed too!
print(a is b)    # 'is' checks whether two names refer to the same object
```

To get an independent list, make a **copy**:

```python
a = [1, 2, 3]
b = a.copy()     # or list(a), or a[:]
b.append(4)
print("a:", a, " b:", b)
print(a is b, a == b)   # different objects; '==' compares contents
```

A normal copy is **shallow**: it copies the outer list, but nested objects are still shared. Use `copy.deepcopy` for nested structures:

```python
import copy

grid = [[0, 0], [0, 0]]
shallow = grid.copy()
deep = copy.deepcopy(grid)

grid[0][0] = 99
print("shallow:", shallow)   # affected - inner lists are shared
print("deep:   ", deep)      # not affected
```

The same issue appears when "multiplying" nested lists:

```python
bad = [[0] * 3] * 2      # two references to ONE inner list
bad[0][0] = 1
print(bad)

good = [[0] * 3 for _ in range(2)]   # list comprehension (Module 3) creates separate lists
good[0][0] = 1
print(good)
```

This topic will return with NumPy arrays and pandas DataFrames, where *views* versus *copies* matter a lot.

## 2.4 Tuples

A tuple is an **ordered, immutable** sequence. Use it for data that should not change.

```python
shape = (150, 4)
print(shape[0], len(shape))

try:
    shape[0] = 200
except TypeError as e:
    print("TypeError:", e)

single = (5,)          # a one-element tuple needs a trailing comma
not_a_tuple = (5)      # this is just the integer 5
print(type(single), type(not_a_tuple))

point = 3, 4           # parentheses are optional
x, y = point           # tuple unpacking
print(x, y)
```

Why use tuples instead of lists?

* **Safety:** nobody can accidentally modify them.
* **Hashability:** tuples (of immutable items) can be dictionary keys or set elements; lists cannot.
* **Convention:** functions often return several values as a tuple (Module 4).

## 2.5 Sets

A set is an **unordered collection of unique** items. Membership testing is extremely fast.

```python
labels = ["cat", "dog", "cat", "bird", "dog", "cat"]
classes = set(labels)
print(classes)
print(len(classes), "unique classes")

classes.add("fish")
classes.discard("bird")     # remove if present, no error if absent
print(classes)
print("cat" in classes)

empty_set = set()           # NOTE: {} creates an empty DICTIONARY, not a set
```

Set algebra is useful, for example to compare vocabularies or check for overlap between training and test IDs:

```python
train_ids = {1, 2, 3, 4, 5}
test_ids = {4, 5, 6, 7}
print(train_ids & test_ids)   # intersection - data leakage!
print(train_ids | test_ids)   # union
print(train_ids - test_ids)   # difference
print(train_ids ^ test_ids)   # symmetric difference
print({1, 2} <= train_ids)    # subset test
```

## 2.6 Dictionaries

A dictionary maps **keys** to **values**. Keys must be unique and hashable (strings, numbers, tuples). Since Python 3.7, dictionaries keep insertion order.

```python
params = {
    "learning_rate": 0.01,
    "n_estimators": 100,
    "max_depth": None,
    "criterion": "gini",
}
print(params["learning_rate"])

params["n_estimators"] = 200     # update a value
params["random_state"] = 42      # add a new key
del params["criterion"]          # remove a key
print(params)
print(len(params), "max_depth" in params)   # 'in' checks KEYS
```

Accessing a missing key raises a `KeyError`; `.get()` lets you provide a default instead:

```python
try:
    params["momentum"]
except KeyError as e:
    print("KeyError:", e)

print(params.get("momentum"))         # None
print(params.get("momentum", 0.9))    # default value
```

### Views: keys, values, items

```python
print(list(params.keys()))
print(list(params.values()))
for key, value in params.items():     # the most common way to loop over a dict
    print(f"{key:>15}: {value}")
```

### Building, merging and updating

```python
defaults = {"lr": 0.1, "epochs": 10, "batch": 32}
user = {"lr": 0.01, "dropout": 0.5}

merged = defaults | user          # Python 3.9+: values on the right win
print(merged)

config = dict(defaults)           # copy
config.update(user)               # in-place merge
print(config)

pairs = [("a", 1), ("b", 2)]
print(dict(pairs))
print(dict(zip(["x", "y"], [10, 20])))   # zip pairs items from two sequences
```

### Counting with dictionaries

Counting occurrences is a classic task (e.g. class distribution, word frequencies):

```python
labels = ["cat", "dog", "cat", "bird", "dog", "cat"]
counts = {}
for label in labels:
    counts[label] = counts.get(label, 0) + 1
print(counts)

# The standard library has a ready-made tool for this:
from collections import Counter
c = Counter(labels)
print(c)
print(c.most_common(1))
```

`collections.defaultdict` creates a default value automatically for missing keys:

```python
from collections import defaultdict

by_class = defaultdict(list)
samples = [("cat", 1.2), ("dog", 3.4), ("cat", 0.9)]
for label, value in samples:
    by_class[label].append(value)
print(dict(by_class))
```

## 2.7 Nested structures

## 2.8 Choosing a data structure — and why speed matters

Checking `x in my_list` scans the list item by item (time grows with length: **O(n)**). Sets and dictionaries use **hash tables**, so `x in my_set` takes roughly constant time (**O(1)**).

```python
import time

n = 1_000_000            # underscores make large numbers readable
big_list = list(range(n))
big_set = set(big_list)

start = time.perf_counter()
for _ in range(100):
    (n - 1) in big_list
t_list = time.perf_counter() - start

start = time.perf_counter()
for _ in range(100):
    (n - 1) in big_set
t_set = time.perf_counter() - start

print(f"list: {t_list:.4f} s   set: {t_set:.6f} s   -> set is ~{t_list / t_set:,.0f}x faster")
```

Rules of thumb:

* Need order and will modify? → **list**
* Fixed record that should not change? → **tuple**
* Need uniqueness or fast membership tests? → **set**
* Need to look things up by name/key? → **dict**

## Summary

* Lists are ordered and mutable; tuples are ordered and immutable.
* Sets store unique items and support fast membership tests and set algebra.
* Dictionaries map keys to values; use `.get()` for safe access and `.items()` to loop.
* Assignment copies **references**, not objects. Use `.copy()` or `copy.deepcopy()` when you need independent data.
